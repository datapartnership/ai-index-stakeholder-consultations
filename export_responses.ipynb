# Download survey responses

Downloads every response from the Azure backend as `survey-responses.csv` and loads it into a table.

The export key is **not stored in this notebook**. Set it in the `SURVEY_EXPORT_KEY` environment variable, or paste it when prompted. `survey-responses.csv` is git-ignored because it contains respondents' answers.

```python
pip install certifi
```

```python
import getpass
import os
import ssl
import urllib.error
import urllib.request
from pathlib import Path

import certifi

EXPORT_URL = "https://stakeholdersurvey-hmgfbshjdzh9egg2.eastus-01.azurewebsites.net/api/export"
OUTPUT = Path("survey-responses.csv")

# Use certifi's CA bundle; macOS python.org builds often lack system root certs.
ssl_context = ssl.create_default_context(cafile=certifi.where())

key = os.environ.get("SURVEY_EXPORT_KEY") or getpass.getpass("Export key: ")
request = urllib.request.Request(EXPORT_URL, headers={"X-Export-Key": key})
try:
    with urllib.request.urlopen(request, timeout=60, context=ssl_context) as response:
        OUTPUT.write_bytes(response.read())
except urllib.error.HTTPError as err:
    raise RuntimeError(f"Export failed: HTTP {err.code} {err.reason}") from None

print(f"Saved {OUTPUT.resolve()} ({OUTPUT.stat().st_size:,} bytes)")
```
